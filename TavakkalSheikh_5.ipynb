```python
#Task5: Sales Data Analysis
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\DELL\Desktop\Superstore.csv", encoding="latin1")
print("Dataset Preview")
print(df.head())
```

```python
# TASK 1 : MONTHLY REVENUE TREND
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Month_Year"] = df["Order Date"].dt.to_period("M")
monthly_revenue = df.groupby("Month_Year")["Sales"].sum()
plt.figure(figsize=(10,5))
plt.plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values
)
plt.title("Monthly Revenue Trend")
plt.xlabel("Month-Year")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
# TASK 2 : TOP SELLING PRODUCTS
top_products = (
    df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)
)
plt.figure(figsize=(10,5))
plt.bar(
    top_products.index,
    top_products.values
)
plt.title("Top 10 Selling Products")
plt.xlabel("Product Name")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.show()
```

```python
# TASK 3 : PROFIT ANALYSIS
profit_category = (
    df.groupby("Category")["Profit"].sum()
)
plt.figure(figsize=(6,4))
plt.bar(
    profit_category.index,
    profit_category.values
)
plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.show()
```

```python
# BUSINESS INSIGHTS
print("\nBusiness Insights")
print("\nHighest Revenue Month:")
print(monthly_revenue.idxmax())
print("\nTop Selling Product:")
print(top_products.idxmax())
print("\nMost Profitable Category:")
print(profit_category.idxmax())
```
